```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# a

#np.random.seed(12)

def experiment():
  urna = [0]*3 + [1]*4 + [2]*2 # 0 pentru rosu, 1 pentru albastru, 2 pentru negru
  zar = np.random.randint(1, 7)

  if zar in (2, 3, 5):
    urna.append(2)
  elif zar == 6:
    urna.append(0)
  else:
    urna.append(1)

  return np.random.choice(urna)


N = 100000
results = []
for i in range(N):
    results.append(experiment())
results = np.array(results)
```

```python
# b

p_rosu = np.mean(results == 0)
print("Probabilitate bila rosie:", p_rosu)
```

```python
# c

# R = bila extrasa este rosie
# P = aruncarea la zar este numar prim, cu P(P) = 3/6 = 1/2
# S = aruncarea la zar este 6, cu P(S) = 1/6
# A = aruncarea la zar este ce ramane (1 sau 4), cu P(A) = 2/6 = 1/3
# P(R) = P(R | P)·P(P) + P(R | S)·P(S) + P(R | A)·P(A)
# P(R) = 3/10 · 1/2 + 4/10 · 1/6 + 3/10 · 1/3 = 19/60 = aprox 0,3167

p_teoretic = 19/60
print("Probabilitatea teoretica:", p_teoretic)
print("Probabilitatea simulata:", p_rosu)
print("Diferenta intre probabilitati:", abs((p_rosu - p_teoretic)))
```
