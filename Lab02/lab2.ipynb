```python
import numpy as np
```

```python
# ex2

N = 100000
nr_pozitiv = 0 # au testul pozitiv
nr_bolnavi_pozitivi = 0 # au testul pozitiv si sunt bolnavi

for i in range(N):
    if np.random.rand() < 0.01:
        bolnav = True
    else:
        bolnav = False

    if bolnav:
        pozitiv = np.random.rand() < 0.95
    else:
        pozitiv = np.random.rand() < 0.10

    if pozitiv:
        nr_pozitiv = nr_pozitiv + 1
        if bolnav:
            nr_bolnavi_pozitivi = nr_bolnavi_pozitivi + 1

p_simulat = nr_bolnavi_pozitivi / nr_pozitiv
p_teoretic = 0.0095 / 0.1085

print("Simulat:", p_simulat)
print("Teoretic:", p_teoretic)
```
