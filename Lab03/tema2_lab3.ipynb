**Ex. 1** Avem 36 de cazuri egal probabile $(X, Y)$.

**1.** $X$ par are 18 cazuri. Dintre ele, suma este $\ge 10$ pentru $(4,6), (6,4), (6,5), (6,6)$, deci 4 cazuri:

$$P(X+Y \ge 10 \mid X \text{ par}) = \frac{4}{18} = \frac29 \approx 0.222$$

**2.** Cel putin un zar par: $1 - \left(\tfrac12\right)^2 = \tfrac34$, adica 27 de cazuri. Suma este $\ge 10$ in 6 cazuri: $(4,6), (5,5), (6,4), (5,6), (6,5), (6,6)$. Doar $(5,5)$ nu are niciun zar par, deci raman 5 cazuri:

$$P(X+Y \ge 10 \mid \text{cel putin un zar par}) = \frac{5}{27} \approx 0.185$$

**3.** Suma 8 apare in 5 cazuri: $(2,6), (3,5), (4,4), (5,3), (6,2)$. Ambele zaruri sunt pare in 3 dintre ele:

$$P(\text{ambele pare} \mid X+Y = 8) = \frac{3}{5} = 0.6$$

**4.** $P(A) = \frac{6}{36} = \frac16$, $P(B) = \frac12$, $P(A \cap B) = \frac{4}{36} = \frac19$. Cum $\frac19 \ne \frac1{12} = P(A)\,P(B)$, evenimentele **nu sunt independente**. Intuitiv, daca $X$ este par, $X = 6$ este mai probabil, iar 6 duce des la suma mare.

```python
# ex. 1, extensie python
import numpy as np

rng = np.random.default_rng()
n = 100000

X = rng.integers(1, 7, size=n)   # primul zar
Y = rng.integers(1, 7, size=n)   # al doilea zar
suma = X + Y

# evenimentele ca masti booleene
X_par = X % 2 == 0
Y_par = Y % 2 == 0
A = suma >= 10

def conditionata(eveniment, conditie):
    # P(eveniment | conditie) = cazurile care le au pe amandoua / cazurile cu conditia
    return (eveniment & conditie).sum() / conditie.sum()

# (estimare din simulare, valoare exacta)
rezultate = {
    "1. P(S>=10 | X par)": (conditionata(A, X_par), 4 / 18),
    "2. P(S>=10 | cel putin un zar par)": (conditionata(A, X_par | Y_par), 5 / 27),
    "3. P(ambele pare | S=8)": (conditionata(X_par & Y_par, suma == 8), 3 / 5),
    "4. P(A si B)": ((A & X_par).mean(), 1 / 9),
    "4. P(A) * P(B)": (A.mean() * X_par.mean(), 1 / 12),
}

for nume, (estimat, exact) in rezultate.items():
    print(f"{nume}: estimat = {estimat:.4f}, exact = {exact:.4f}")
```

**Ex. 2** Notam $S$ = mesaj spam, $L$ = mesaj legitim, cu $P(S) = 0.3$, $P(L) = 0.7$.

**1.** Din formula lui Bayes, cu $P(\text{oferta} \mid S) = 0.8$ si $P(\text{oferta} \mid L) = 0.15$:

$$P(S \mid \text{oferta}) = \frac{0.8 \cdot 0.3}{0.8 \cdot 0.3 + 0.15 \cdot 0.7} = \frac{0.24}{0.345} = \frac{16}{23} \approx 0.696$$

**2.** Cu independenta conditionata (dat clasa), probabilitatile cuvintelor se inmultesc:

$$P(\text{oferta}, \text{gratuit} \mid S) = 0.8 \cdot 0.6 = 0.48, \qquad P(\text{oferta}, \text{gratuit} \mid L) = 0.15 \cdot 0.1 = 0.015$$

$$P(S \mid \text{oferta}, \text{gratuit}) = \frac{0.48 \cdot 0.3}{0.48 \cdot 0.3 + 0.015 \cdot 0.7} = \frac{0.144}{0.1545} = \frac{96}{103} \approx 0.932$$

**3.** Ipoteza e nerealista pentru ca cuvintele dintr-un mesaj nu apar independent, nici macar in aceeasi clasa. De exemplu, „oferta" si „gratuit" apar adesea impreuna in aceleasi mesaje promotionale, deci stim deja ceva despre al doilea cuvant cand il vedem pe primul. Inmultind probabilitatile numaram aceeasi informatie de doua ori, iar rezultatul (0.932) este prea increzator.

**4.** Regula de clasificare (Naive Bayes): pentru un mesaj cu cuvintele $w_1, \dots, w_k$, calculam

$$P(S \mid w_1, \dots, w_k) = \frac{P(S) \prod_i P(w_i \mid S)}{P(S) \prod_i P(w_i \mid S) + P(L) \prod_i P(w_i \mid L)}$$

si clasificam mesajul ca **spam** daca $P(S \mid w_1, \dots, w_k) > \tau$, altfel ca legitim. Pragul $\tau = 0.5$ este standard, dar il putem ridica (de exemplu la 0.9) daca vrem sa evitam mesajele legitime marcate gresit ca spam. Cu $\tau = 0.5$, mesajul cu „oferta" (0.696) si cel cu ambele cuvinte (0.932) sunt clasificate ca spam.

**Ex. 3**

**1.** Ratele de promovare:

| Program | Disciplina A | Disciplina B |
|---|---|---|
| Programul 1 | $81/90 = 0.90$ | $19/20 = 0.95$ |
| Programul 2 | $1/10 = 0.10$ | $16/80 = 0.20$ |

**2.** In ambele programe, disciplina **B** are rata de promovare mai mare ($0.95 > 0.90$ si $0.20 > 0.10$).

**3.** Ratele globale:

$$\text{A: } \frac{81 + 1}{90 + 10} = \frac{82}{100} = 0.82, \qquad \text{B: } \frac{19 + 16}{20 + 80} = \frac{35}{100} = 0.35$$

**4.** Ordinea se inverseaza pentru ca programul este o variabila de confuzie, iar grupurile au marimi foarte diferite. La disciplina A sunt evaluati in mare parte studenti din Programul 1 (90 din 100), care promoveaza usor la ambele discipline. La disciplina B sunt evaluati in mare parte studenti din Programul 2 (80 din 100), unde ratele sunt mici. Ratele globale sunt medii ponderate cu ponderi diferite:

$$\text{A: } 0.9 \cdot 0.9 + 0.1 \cdot 0.1 = 0.82, \qquad \text{B: } 0.2 \cdot 0.95 + 0.8 \cdot 0.2 = 0.35$$

**5.** Notam $P$ = student promovat, $D$ = disciplina, $G$ = programul. Fenomenul este:

$$P(P \mid B, G_i) > P(P \mid A, G_i) \ \text{ pentru } i = 1, 2, \qquad \text{dar} \qquad P(P \mid B) < P(P \mid A)$$

Explicatia vine din formula probabilitatii totale, $P(P \mid D) = \sum_i P(P \mid D, G_i)\, P(G_i \mid D)$, in care ponderile $P(G_i \mid D)$ sunt diferite pentru A si pentru B.

```python
# ex. 3, extensie python: exemplu propriu de paradox simpson
# comparam doua reclame (X si Y) dupa rata de conversie, pe tipul de dispozitiv
import pandas as pd

df = pd.DataFrame({
    "reclama":   ["X", "X", "Y", "Y"],
    "dispozitiv": ["mobil", "desktop", "mobil", "desktop"],
    "conversii": [45, 35, 2, 285],
    "vizite":    [900, 100, 50, 950],
})

# rata de conversie pentru fiecare reclama si dispozitiv
df["rata"] = df["conversii"] / df["vizite"]
print(df)

# pe fiecare dispozitiv, reclama X are rata mai mare decat Y
print(df.pivot(index="dispozitiv", columns="reclama", values="rata"))

# rata globala: adunam conversiile si vizitele inainte sa impartim
global_ = df.groupby("reclama")[["conversii", "vizite"]].sum()
global_["rata"] = global_["conversii"] / global_["vizite"]
print(global_)

# globala, Y are rata mai mare, desi X e mai bun pe fiecare dispozitiv:
# X e afisata in mare parte pe mobil (conversie mica), Y pe desktop (conversie mare)
```
